```python
import cv2
import numpy as np
```

```python
image1 = cv2.imread("fundamental_img1.png")
image2 = cv2.imread("fundamental_img2.png")
```

```python
gray1 = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY)
```

```python
sift = cv2.SIFT.create()

keypoints_left, descriptors_left = sift.detectAndCompute(gray1, None)
keypoints_right, descriptors_right = sift.detectAndCompute(gray2, None)

brute_force_matcher = cv2.BFMatcher()
matches = brute_force_matcher.match(descriptors_left, descriptors_right)

matches = sorted(matches, key=lambda x: x.distance)

num_best_matches = 2000

points1 = []
points2 = []

for m in matches[:num_best_matches]:
    points1.append(keypoints_left[m.queryIdx].pt)
    points2.append(keypoints_right[m.trainIdx].pt)

points1 = np.int32(points1)
points2 = np.int32(points2)

F, mask = cv2.findFundamentalMat(points1, points2)
```
