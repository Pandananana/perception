```python
import numpy as np
import cv2
```

```python
image1 = cv2.imread("optflow1.jpg")
image2 = cv2.imread("optflow2.jpg")
```

```python
gray1 = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY)


features = cv2.goodFeaturesToTrack(
    gray1, maxCorners=100, qualityLevel=0.3, minDistance=7
)

features2, status, err = cv2.calcOpticalFlowPyrLK(gray1, gray2, features, None)

largest_displacements = np.mean(abs(features - features2))
```
