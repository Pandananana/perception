# PCA explained variance
Now perform principal component analysis on the given data. 

How many components do you need to express at least 75% of the variance?

recall you call PCA:

```{python}
pca = PCA()
pca.fit(X)
```

Documentation: `help(PCA)` or `%pdoc PCA`


````python
import numpy as np
from sklearn.decomposition import PCA

#clusters.txt located in the clustering folder
X = np.loadtxt('clusters.txt')
````

````python
pca = PCA(0.75)

pca.fit(X)

print(f"Number of components: {pca.n_components_}")
````
