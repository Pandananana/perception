````python
import matplotlib.pyplot as plt  
from sklearn.cluster import KMeans 
from sklearn.metrics import silhouette_score
import numpy as np

# Loading the data
X = np.loadtxt('clusters.txt', dtype=np.float32)
````

# KMeans clustering and silhouette score analysis
Given the dataset `clusters.txt` that has been read into the variable X, find how many clusters are optimal for this dataset using the silhouette score for different numbers of $n$.


Recall you call Kmeans:
```{python}
km = KMeans(n_clusters = n)
km.fit(X)
```

You can call the documentation for KMeans by either doing `help(KMeans)` or `%pdoc KMeans`

Note the following from the silhouette documentation:

```
Note that Silhouette Coefficient is only defined if number of labels
is ``2 <= n_labels <= n_samples - 1``.
```

You can read the documentation for the function using `help(silhouette_score)`
The call signature for silhoutte score is as follows:

````python
scores = []
for k in range(3, 20):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    scores.append(km.inertia_)

plt.plot(range(3, 20), scores)
plt.show()
````

````python
# silhouet plot
for i in range(3, 20):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(X)
    labels = kmeans.predict(X)
    # print(f"Silhouette score for {i} clusters: {silhouette_score(X, labels)}")
    plt.bar(i, silhouette_score(X, labels), color='blue')
````
