```python
import cv2
from matplotlib import pyplot as plt

img = cv2.imread("data/images/circles.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 100, 200)
plt.imshow(edges, cmap="gray")
```

```python
circles = cv2.HoughCircles(edges, method=cv2.HOUGH_GRADIENT, dp=1, minDist=20, param1=50, param2=30, minRadius=10, maxRadius=50)

r = img.copy()
for c in circles[0]:
    center = (int(c[0]), int(c[1]))
    radius = int(c[2])
    r = cv2.circle(r, center, radius, (0, 0, 255), 5)
    
plt.imshow(r)
print("Number of circles: ", circles.shape[1])
```
