Mean Shift is a mode-seeking algorithm that doesn't require you to specify the number of clusters beforehand. Instead, it initializes by treating each data point as its own cluster center initially, then iteratively shifts these centers toward regions of higher density until they converge.
In this case, each pixel is a data point

![image](attachments/3_6.png)

Each pixel is represented as a feature vector that includes:
Geometric information (spatial coordinates):

x coordinate (1 dimension)
y coordinate (1 dimension)

Color information (for a color image, typically RGB):

Red channel (1 dimension)
Green channel (1 dimension)
Blue channel (1 dimension)

So each pixel becomes a 5-dimensional data point: (x, y, R, G, B)

```python
from sklearn.cluster import KMeans
from matplotlib import pyplot as plt
import numpy as np

X = np.loadtxt("attachments/clusters.txt")

x = []
y = []
for n_clusters in range (1,15):
    km = KMeans(n_clusters = n_clusters)
    km.fit(X)
    inertia = km.inertia_
    x.append(n_clusters)
    y.append(inertia)

plt.plot(x, y)
```
