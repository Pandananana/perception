L3, Stereo Vision, page 58

1. (Least demanding): Dense Local Stereo Vision Algorithm
Local methods use a fixed window around each pixel and compute matching costs independently. This is essentially O(W·H·D) where W×H is image size and D is disparity range—very parallelizable and fast.
2. (Middle): Dense Dynamic Programming-based Stereo Vision Algorithm
DP methods optimize along scanlines (1D optimization), enforcing ordering constraints row by row. Complexity is roughly O(W·H·D²) but still tractable since it avoids full 2D global optimization.
3. (Most demanding): Dense Graph Cuts-based Stereo Vision Algorithm
Graph cuts perform global 2D optimization over the entire image, minimizing an energy function with data terms and smoothness terms. This involves iterative min-cut/max-flow computations, making it significantly more expensive—though it typically produces higher quality results.

Disparity is defined as the difference in horizontal position between corresponding pixels in the left and right images (d = x_left - x_right in a rectified stereo pair).
A low disparity means there's a small horizontal shift between the matching pixels—they appear at nearly the same location in both images. This corresponds to points that are far away from the cameras (think of how distant objects barely shift when you close one eye then the other).
Conversely, high disparity means large positional difference, which indicates objects close to the camera.
