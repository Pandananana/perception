The PnP solver gives you the **camera pose** relative to world coordinates:
- `rvec` → rotation from world to camera frame
- `tvec` → translation from world to camera frame

The transformation from **world to camera** is:

$$P_c = R \cdot P_w + t$$

Here, `t` represents **where the world origin appears** when viewed from the camera's perspective. It's not the camera's position in the world—it's the offset needed to express world points in camera coordinates after rotation.

To go **camera to world**, we invert this:

$$P_w = R^T \cdot (P_c - t)$$

We subtract `t` first to undo the translation, then apply the inverse rotation (R transposed) to get back to world coordinates.
